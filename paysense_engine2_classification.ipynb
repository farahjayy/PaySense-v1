## 2. Generate Synthetic BNPL Dataset (Literature-Grounded v2)
2,000 synthetic Gen-Z Malaysian student/early-career profiles.

**Key distributions anchored to Malaysian literature:**
- **Age** — Tan et al. (2026): 13.3% aged 18–19, 61.3% aged 21–23, 25.3% aged 24–26
- **Employment** — Osman et al. (2024): 61.4% student, 26.4% employed, 7.0% self-employed, 5.2% unemployed
- **Monthly income** — Osman et al. (2024): 65.9% below RM1,000, 15.9% RM1,001–2,999, 15.7% RM3,000–4,999, 2.5% RM5,000+

Features without literature basis (monthly expenses, BNPL usage counts, missed payments) are retained as domain-informed assumptions, clearly flagged in code.

```python
N = 2000
rng = np.random.default_rng(SEED)

# ══════════════════════════════════════════════════════════════════════════════
# LITERATURE-GROUNDED SYNTHETIC DATA GENERATION
# PaySense Engine 2 — v2 (literature-anchored)
#
# Sources used to parameterise distributions:
#   [1] Tan et al. (2026) — BNPL Usage & Debt Management Among Malaysian
#       University Students  (N=150, active BNPL users)
#   [2] Osman et al. (2024) — How BNPL is Shaping Gen Z's Spending Spree
#       in Malaysia  (N=440, Gen Z aware of BNPL)
#
# Feature-by-feature grounding status:
#   age               ✅ Literature-derived  — Tan et al. (2026) Table 1
#   employment_status ✅ Literature-derived  — Osman et al. (2024) Table 1
#   monthly_income    ✅ Literature-derived  — Osman et al. (2024) Table 1
#   monthly_expenses  ⚠️ Assumption          — no Malaysian paper reports this
#   num_bnpl_plans    ⚠️ Assumption          — no Malaysian paper reports this
#   bnpl_outstanding  ⚠️ Assumption          — no Malaysian paper reports this
#   missed_payments   ⚠️ Assumption          — no Malaysian paper reports this
#   forecasted_cf     ⚠️ Assumption          — Engine 1 output, simulated here
#   (derived ratios)  ⚠️ Computed from above
#   bnpl_risk         ⚠️ Assumption          — risk label logic, no paper benchmark
# ══════════════════════════════════════════════════════════════════════════════

# ── AGE ───────────────────────────────────────────────────────────────────────
# Source: Tan et al. (2026) Table 1 — age distribution of Malaysian university
# students who actively use BNPL (N=150):
#   Below 20 : 13.3%  →  mapped to 18-19
#   21-23    : 61.3%
#   24-26    : 25.3%
#   Above 27 :  0.0%  (excluded — not present in either paper's sample)
# Cap at 26; no respondent above 27 in Tan et al.; Osman et al. confirms 13-26 range.
age_p = np.array([0.133, 0.613, 0.253]); age_p = age_p / age_p.sum()  # normalise: sums to 0.999 as literally transcribed
age_brackets = rng.choice([0, 1, 2], N, p=age_p)  # 0=18-19, 1=21-23, 2=24-26
age = np.where(age_brackets == 0,
               rng.integers(18, 20, N),
     np.where(age_brackets == 1,
               rng.integers(21, 24, N),
               rng.integers(24, 27, N)))

# ── EMPLOYMENT STATUS ─────────────────────────────────────────────────────────
# Source: Osman et al. (2024) Table 1 — occupation of Gen Z BNPL-aware sample (N=440):
#   Student       : 61.4%
#   Employed      : 26.4%
#   Self-Employed :  7.0%
#   Unemployed    :  5.2%
# Encoding: 0=student, 1=employed, 2=self-employed, 3=unemployed
employment = rng.choice([0, 1, 2, 3], N, p=[0.614, 0.264, 0.070, 0.052])

# ── MONTHLY INCOME ────────────────────────────────────────────────────────────
# Source: Osman et al. (2024) Table 1 — monthly income of Gen Z BNPL-aware sample:
#   Below RM1,000       : 65.9%
#   RM1,001 – RM2,999   : 15.9%
#   RM3,000 – RM4,999   : 15.7%
#   RM5,000+            :  2.5%
# Bracket assignment is conditioned on employment to maintain realism
# (students skew toward below RM1,000; employed skew toward higher brackets).
# Within each bracket, income is drawn from a uniform sub-range.
income_bracket = np.zeros(N, dtype=int)

for i in range(N):
    emp = employment[i]
    if emp == 0:    # student → heavily skewed to below RM1,000
        income_bracket[i] = rng.choice([0, 1, 2, 3], p=[0.88, 0.10, 0.015, 0.005])
    elif emp == 1:  # employed → spread across brackets
        income_bracket[i] = rng.choice([0, 1, 2, 3], p=[0.30, 0.35, 0.30, 0.05])
    elif emp == 2:  # self-employed → mid brackets
        income_bracket[i] = rng.choice([0, 1, 2, 3], p=[0.40, 0.35, 0.20, 0.05])
    else:           # unemployed → almost all below RM1,000
        income_bracket[i] = rng.choice([0, 1, 2, 3], p=[0.95, 0.04, 0.01, 0.00])

# Overall bracket marginals self-correct toward Osman et al. Table 1 via the
# employment weights (61.4% students who earn mostly below RM1,000 dominate).
monthly_income = np.where(income_bracket == 0,
                    rng.uniform(300, 999, N),
                np.where(income_bracket == 1,
                    rng.uniform(1001, 2999, N),
                np.where(income_bracket == 2,
                    rng.uniform(3000, 4999, N),
                    rng.uniform(5000, 8000, N))))
monthly_income = np.round(monthly_income, 2)

# ── MONTHLY EXPENSES ──────────────────────────────────────────────────────────
# ⚠️ ASSUMPTION — neither paper reports monthly expense data.
# Modelled as a fraction of income; ratio drawn from Uniform(0.60, 1.30).
# Students on low incomes tend toward higher expense ratios (spending > income
# is common when relying on family allowances or PTPTN loans).
expense_ratio    = rng.uniform(0.60, 1.30, N)
monthly_expenses = np.round(monthly_income * expense_ratio, 2)

# ── BNPL USAGE ────────────────────────────────────────────────────────────────
# ⚠️ ASSUMPTION — neither paper reports number of active BNPL plans or balances.
# num_bnpl_plans : integer 0–6 (uniform)
# bnpl_outstanding: proportional to num_bnpl_plans × per-plan amount
num_bnpl_plans   = rng.integers(0, 7, N)
bnpl_outstanding = np.round(num_bnpl_plans * rng.uniform(50, 400, N), 2)

# ── MISSED PAYMENTS ───────────────────────────────────────────────────────────
# ⚠️ ASSUMPTION — neither paper reports missed payment rates.
# Base rate 0–6, then adjusted upward for high BNPL financial stress.
missed_payments = rng.integers(0, 7, N)
missed_payments = np.clip(
    missed_payments + (bnpl_outstanding / np.maximum(monthly_income, 1) * 1.5).astype(int),
    0, 10
).astype(int)

# ── ENGINE 1 LINKAGE: simulated forecasted net cash flow ─────────────────────
# ⚠️ ASSUMPTION — simulates what Engine 1 would forecast for each profile.
noise        = rng.normal(0, 150, N)
forecasted_cf = np.round(
    (monthly_income - monthly_expenses) * rng.uniform(0.7, 1.1, N) + noise, 2
)

# ── DERIVED RATIOS ────────────────────────────────────────────────────────────
income_expense_ratio = np.round(monthly_income / np.maximum(monthly_expenses, 1), 4)
bnpl_income_ratio    = np.round(bnpl_outstanding / np.maximum(monthly_income, 1), 4)
savings_rate         = np.round(
    (monthly_income - monthly_expenses) / np.maximum(monthly_income, 1), 4
)

# ── TARGET: BNPL RISK (0=low, 1=high) ────────────────────────────────────────
# ⚠️ ASSUMPTION — risk label logic; no paper provides binary risk benchmarks.
#
# v3 — continuous scaling (was: hard binary thresholds in v1/v2).
# Found via a live Risk Checker case: a purchase using 115% of monthly income
# scored "safe" (P(risk)=0.08) because missed_payments only counted at >=2
# (1 missed payment == 0 missed payments to the old label) and
# bnpl_income_ratio was a single +2 flag regardless of being 51% or 500% of
# income. See docs/KNOWN_LIMITATIONS.md for the full trace (SHAP-verified).
# Each term below now scales linearly with severity instead of a hard cutoff;
# the overall calibration point (score - 3.5) is kept the same as v1/v2 so the
# resulting class balance is comparable.
missed_component  = np.clip(missed_payments, 0, 6) / 6 * 4
ratio_component   = np.clip(bnpl_income_ratio, 0, 3) / 3 * 4
savings_component = np.clip(-savings_rate, 0, 1) * 2
ier_component     = np.clip(0.9 - income_expense_ratio, 0, 0.9) / 0.9 * 1
cf_component      = np.clip(-forecasted_cf - 100, 0, 1000) / 1000 * 2
plans_component   = np.clip(num_bnpl_plans, 0, 6) / 6 * 1

risk_score = (
    missed_component + ratio_component + savings_component
    + ier_component + cf_component + plans_component
)
risk_prob = 1 / (1 + np.exp(-(risk_score - 3.5)))
bnpl_risk = (rng.uniform(0, 1, N) < risk_prob).astype(int)

df = pd.DataFrame({
    'age':                  age,
    'employment_status':    employment,
    'monthly_income':       monthly_income,
    'monthly_expenses':     monthly_expenses,
    'num_bnpl_plans':       num_bnpl_plans,
    'bnpl_outstanding':     bnpl_outstanding,
    'missed_payments':      missed_payments,
    'forecasted_cash_flow': forecasted_cf,
    'income_expense_ratio': income_expense_ratio,
    'bnpl_income_ratio':    bnpl_income_ratio,
    'savings_rate':         savings_rate,
    'bnpl_risk':            bnpl_risk
})

# ── LITERATURE ALIGNMENT CHECK ────────────────────────────────────────────────
print("=" * 60)
print("  LITERATURE ALIGNMENT REPORT")
print("=" * 60)
print(f"\n[age] Range: {age.min()}–{age.max()}  (literature: 18–26)")
print(f"  18-19 : {(age<=19).mean()*100:.1f}%  (target 13.3%)")
print(f"  21-23 : {((age>=21)&(age<=23)).mean()*100:.1f}%  (target 61.3%)")
print(f"  24-26 : {(age>=24).mean()*100:.1f}%  (target 25.3%)")

emp_labels = {0:'student', 1:'employed', 2:'self-employed', 3:'unemployed'}
emp_targets = {0:61.4, 1:26.4, 2:7.0, 3:5.2}
print(f"\n[employment_status]  (Osman et al. 2024 targets in brackets)")
for code, label in emp_labels.items():
    actual = (employment==code).mean()*100
    target = emp_targets[code]
    print(f"  {label:<15}: {actual:.1f}%  (target {target}%)")

print(f"\n[monthly_income]  (Osman et al. 2024 targets in brackets)")
print(f"  Below RM1,000   : {(monthly_income<1000).mean()*100:.1f}%  (target 65.9%)")
print(f"  RM1,001-2,999   : {((monthly_income>=1001)&(monthly_income<=2999)).mean()*100:.1f}%  (target 15.9%)")
print(f"  RM3,000-4,999   : {((monthly_income>=3000)&(monthly_income<=4999)).mean()*100:.1f}%  (target 15.7%)")
print(f"  RM5,000+        : {(monthly_income>=5000).mean()*100:.1f}%  (target  2.5%)")
print(f"  Mean income     : RM{monthly_income.mean():.0f}")

print(f"\n[bnpl_risk]  Class distribution:")
print(f"  Low Risk  (0): {(bnpl_risk==0).sum()}  ({(bnpl_risk==0).mean()*100:.1f}%)")
print(f"  High Risk (1): {(bnpl_risk==1).sum()}  ({(bnpl_risk==1).mean()*100:.1f}%)")

print(f"\n[features with NO literature basis — kept as domain assumptions]:")
print(f"  monthly_expenses, num_bnpl_plans, bnpl_outstanding,")
print(f"  missed_payments, forecasted_cash_flow, derived ratios, bnpl_risk")
print("=" * 60)
print(f"\nDataset: {df.shape[0]} rows x {df.shape[1]} columns")
df.head()
```

## 4. Feature Engineering

```python
FEATURES = [
    'age', 'employment_status',
    'monthly_income', 'monthly_expenses',
    'num_bnpl_plans', 'bnpl_outstanding', 'missed_payments',
    'forecasted_cash_flow',
    'income_expense_ratio', 'bnpl_income_ratio', 'savings_rate'
]
TARGET = 'bnpl_risk'

X = df[FEATURES].copy()
y = df[TARGET].copy()

print(f"Features : {len(FEATURES)}")
print(f"Samples  : {len(X)}")
print(f"\nFeature list:")
for i, f in enumerate(FEATURES, 1):
    print(f"  {i:2d}. {f}")
```

## 5. Train / Val / Test Split + SMOTE
- Split: 70% train, 15% validation, 15% test (stratified)
- SMOTE applied on **train set only** — val and test remain untouched (real-world distribution)

```python
# 70 / 15 / 15 stratified split
X_temp, X_test,  y_temp, y_test  = train_test_split(X, y, test_size=0.15, random_state=SEED, stratify=y)
X_train, X_val,  y_train, y_val  = train_test_split(X_temp, y_temp, test_size=0.15/0.85,
                                                     random_state=SEED, stratify=y_temp)
print(f"Train : {len(X_train):4d} rows  |  High-risk: {y_train.sum()} ({y_train.mean()*100:.1f}%)")
print(f"Val   : {len(X_val):4d} rows  |  High-risk: {y_val.sum()} ({y_val.mean()*100:.1f}%)")
print(f"Test  : {len(X_test):4d} rows  |  High-risk: {y_test.sum()} ({y_test.mean()*100:.1f}%)")
```

```python
# SMOTE on train only
smote = SMOTE(random_state=SEED, k_neighbors=5)
X_train_sm, y_train_sm = smote.fit_resample(X_train, y_train)

print(f"\nAfter SMOTE — Train: {len(X_train_sm)} rows")
print(f"  Low Risk  (0): {(y_train_sm==0).sum()}")
print(f"  High Risk (1): {(y_train_sm==1).sum()}")

fig, axes = plt.subplots(1, 2, figsize=(10, 4))
for ax, counts, title in zip(axes,
    [y_train.value_counts(), y_train_sm.value_counts()],
    ['Before SMOTE', 'After SMOTE']):
    ax.bar(['Low Risk','High Risk'], [counts.get(0,0), counts.get(1,0)],
           color=['#2ecc71','#e74c3c'], alpha=0.8)
    ax.set_title(title); ax.set_ylabel('Count')
plt.suptitle('SMOTE Effect on Training Set', fontweight='bold')
plt.tight_layout()
plt.savefig(MODEL_DIR + 'smote_effect.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
def evaluate_classifier(model, X_val, y_val, X_test, y_test, model_name):
    y_val_pred   = model.predict(X_val)
    y_test_pred  = model.predict(X_test)
    y_test_prob  = model.predict_proba(X_test)[:, 1]

    acc  = accuracy_score(y_test, y_test_pred)
    prec = precision_score(y_test, y_test_pred, zero_division=0)
    rec  = recall_score(y_test, y_test_pred, zero_division=0)
    f1   = f1_score(y_test, y_test_pred, zero_division=0)
    auc  = roc_auc_score(y_test, y_test_prob)

    print(f"\n{'='*48}  {model_name}")
    print(f"  Accuracy  : {acc:.4f}")
    print(f"  Precision : {prec:.4f}")
    print(f"  Recall    : {rec:.4f}   ← prioritised")
    print(f"  F1        : {f1:.4f}   ← primary selection metric")
    print(f"  ROC-AUC   : {auc:.4f}   ← primary selection metric")
    print(f"\n{classification_report(y_test, y_test_pred, target_names=['Low Risk','High Risk'])}")

    # Confusion matrix
    cm = confusion_matrix(y_test, y_test_pred)
    fig, axes = plt.subplots(1, 2, figsize=(12, 4))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', ax=axes[0],
                xticklabels=['Low','High'], yticklabels=['Low','High'])
    axes[0].set_title(f'{model_name} — Confusion Matrix')
    axes[0].set_ylabel('Actual'); axes[0].set_xlabel('Predicted')

    fpr, tpr, _ = roc_curve(y_test, y_test_prob)
    axes[1].plot(fpr, tpr, linewidth=2, label=f'AUC={auc:.3f}')
    axes[1].plot([0,1],[0,1],'k--', alpha=0.4)
    axes[1].set_xlabel('False Positive Rate')
    axes[1].set_ylabel('True Positive Rate')
    axes[1].set_title(f'{model_name} — ROC Curve')
    axes[1].legend()
    plt.tight_layout()
    plt.savefig(MODEL_DIR + f'{model_name.lower().replace(" ","_")}_eval.png',
                dpi=150, bbox_inches='tight')
    plt.show()

    return {
        'model': model_name,
        'Accuracy':  round(acc, 4),
        'Precision': round(prec, 4),
        'Recall':    round(rec, 4),
        'F1':        round(f1, 4),
        'ROC_AUC':   round(auc, 4),
        'F1_AUC_avg': round((f1 + auc) / 2, 4)
    }
```

## 6. Model 1 — Random Forest

```python
model_rf = RandomForestClassifier(
    n_estimators=200, max_depth=10, min_samples_split=5,
    min_samples_leaf=2, class_weight='balanced',
    random_state=SEED, n_jobs=-1
)
model_rf.fit(X_train_sm, y_train_sm)
print("Random Forest fitted.")
print(f"OOB not enabled (use cross-val for generalisation estimate)")

# 5-fold CV on train set
cv_scores = cross_val_score(model_rf, X_train_sm, y_train_sm, cv=5, scoring='f1', n_jobs=-1)
print(f"5-Fold CV F1: {cv_scores.mean():.4f} ± {cv_scores.std():.4f}")
```

```python
rf_metrics = evaluate_classifier(model_rf, X_val, y_val, X_test, y_test, 'Random Forest')
```

## 7. Model 2 — XGBoost

```python
scale_pos = (y_train_sm == 0).sum() / (y_train_sm == 1).sum()

model_xgb = xgb.XGBClassifier(
    n_estimators=300, max_depth=6, learning_rate=0.05,
    subsample=0.8, colsample_bytree=0.8,
    scale_pos_weight=scale_pos,
    use_label_encoder=False, eval_metric='logloss',
    random_state=SEED, n_jobs=-1, verbosity=0
)
model_xgb.fit(
    X_train_sm, y_train_sm,
    eval_set=[(X_val, y_val)],
    verbose=False
)
print("XGBoost fitted.")

cv_scores_xgb = cross_val_score(model_xgb, X_train_sm, y_train_sm, cv=5, scoring='f1', n_jobs=-1)
print(f"5-Fold CV F1: {cv_scores_xgb.mean():.4f} ± {cv_scores_xgb.std():.4f}")
```

```python
xgb_metrics = evaluate_classifier(model_xgb, X_val, y_val, X_test, y_test, 'XGBoost')
```

## 8. Model 3 — LightGBM

## 10. SHAP Explainability on Winner
SHAP (SHapley Additive exPlanations) shows which features push a prediction toward high risk or low risk.

```python
winner_model = {'Random Forest': model_rf, 'XGBoost': model_xgb, 'LightGBM': model_lgb}[WINNER_NAME]

# Tree explainer works for all three candidates
explainer   = shap.TreeExplainer(winner_model)
shap_values = explainer.shap_values(X_test)

# For binary classifiers some libraries return list [class0, class1]
if isinstance(shap_values, list):
    sv = shap_values[1]   # class 1 = High Risk
else:
    sv = shap_values

print(f"SHAP values computed for {WINNER_NAME}")
print(f"Shape: {sv.shape}  (samples x features)")
```

```python
# ── Global: Feature Importance (mean |SHAP|) ─────────────────────────────────
shap_importance = pd.DataFrame({
    'feature': FEATURES,
    'mean_abs_shap': np.abs(sv).mean(axis=0)
}).sort_values('mean_abs_shap', ascending=True)

fig, ax = plt.subplots(figsize=(9, 6))
colors_bar = ['#e74c3c' if v > shap_importance['mean_abs_shap'].median()
              else '#3498db' for v in shap_importance['mean_abs_shap']]
ax.barh(shap_importance['feature'], shap_importance['mean_abs_shap'],
        color=colors_bar, alpha=0.85)
ax.set_title(f'SHAP Feature Importance — {WINNER_NAME}\n(Mean |SHAP Value| across test set)',
             fontweight='bold')
ax.set_xlabel('Mean |SHAP Value|')
plt.tight_layout()
plt.savefig(MODEL_DIR+'shap_importance.png', dpi=150, bbox_inches='tight')
plt.show()
print(shap_importance.sort_values('mean_abs_shap', ascending=False).to_string(index=False))
```

```python
# ── SHAP Summary (beeswarm) ──────────────────────────────────────────────────
plt.figure(figsize=(10, 7))
shap.summary_plot(sv, X_test, feature_names=FEATURES, show=False, plot_size=None)
plt.title(f'SHAP Summary Plot — {WINNER_NAME}', fontweight='bold', pad=15)
plt.tight_layout()
plt.savefig(MODEL_DIR+'shap_summary.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
# ── SHAP Waterfall for a high-risk prediction ────────────────────────────────
y_test_prob = winner_model.predict_proba(X_test)[:, 1]
high_risk_idx = np.where((winner_model.predict(X_test) == 1) & (y_test_prob > 0.75))[0]

if len(high_risk_idx) > 0:
    sample_idx = high_risk_idx[0]
    print(f"Explaining sample #{sample_idx} — predicted HIGH RISK (prob={y_test_prob[sample_idx]:.3f})")
    print("\nFeature values:")
    for f, v in zip(FEATURES, X_test.iloc[sample_idx].values):
        print(f"  {f:<25}: {v}")

    exp = shap.Explanation(
        values=sv[sample_idx],
        base_values=explainer.expected_value[1] if isinstance(explainer.expected_value, list)
                    else explainer.expected_value,
        data=X_test.iloc[sample_idx].values,
        feature_names=FEATURES
    )
    plt.figure(figsize=(10, 5))
    shap.waterfall_plot(exp, show=False)
    plt.title(f'SHAP Waterfall — High-Risk Profile (prob={y_test_prob[sample_idx]:.3f})',
              fontweight='bold')
    plt.tight_layout()
    plt.savefig(MODEL_DIR+'shap_waterfall_highrisk.png', dpi=150, bbox_inches='tight')
    plt.show()
else:
    print("No high-confidence high-risk samples found in test set.")
```
